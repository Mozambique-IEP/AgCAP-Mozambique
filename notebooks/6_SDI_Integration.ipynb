## Logic and Rationale for Creating Production, Cooling and PUE Potential

This notebook creates a few **new columns** in `settles_gdf_analyzed`:

- `Fresh Markets Accessibility - class`
- `National Market Accessibility - class`
- `Export Market Accessibility - class`
- `Cooling_potential`
- `Irrigation_potential`
- `Combined_PUE_potential` 

The goal is to derive **interpretable, categorical indicators** of productive-use potential per settlement, based on the existing continuous indices in the dataset.

---

### 1. Accessibility_potential

This is the classified version of the accessibility columns to Fresh, National and Export Market(s) - deriving from the core analysis notebook - which we classify into four categories, using **quantiles** (quartiles):
   - Low  
   - Medium  
   - High  
   - Very High  

   This is done with `pandas.qcut`, which splits the distribution into 4 groups based on the actual data (evidence-based thresholds).


### 2. Cooling_potential

**Inputs (cooling-related indices):**

- `Ag Cooling Demand ALL Markets`
- `Fish Cooling Demand ALL Markets`

(All scaled 0–1)

**Rationale:**

Cooling potential reflects **how strong the demand for cooling is**, across:
- Agricultural products
- Fishery products
- Different market types (export / national / fresh)

If cooling demand for **any** of these streams is high, the settlement is a strong candidate for a cold-chain intervention.

**Steps:**

1. For each of the six cooling indices, we again classify values using **quantile-based categories**:
   - Low  
   - Medium  
   - High  
   - Very High  

   This is done separately for each column, so that each indicator uses its own data distribution.

2. For each settlement, we now have up to six class labels (one per cooling index). We convert these labels to ordinal scores (Low=0, Medium=1, High=2, Very High=3) and take the **maximum** across the six.

   - If a settlement is Medium for most, but Very High for Fish Export → Cooling_potential = Very High  
   - If it is Low everywhere → Cooling_potential = Low  

3. We convert this ordinal score back to a label and store the result as:

   - `Cooling_potential` ∈ {Low, Medium, High, Very High}

This ensures that each settlement’s cooling potential reflects **the strongest cooling demand signal** across all ag/fish and market channels.

---

### 3. Irrigation_potential 

**Inputs:**

- `Irrigation: months with (critical) need` (0–12)

**Rationale:**

A settlement can have strong potential for irrigation if the number of "critical" months for water need is high. 

**Steps:**

1. For the `Irrigation: months with (critical) need` column, we classify values into **four evidence-based categories**, using **quantiles** (quartiles):
   - Low  
   - Medium  
   - High  
   - Very High  

   This is done with `pandas.qcut`, which splits the distribution into 4 groups based on the actual data (evidence-based thresholds).

2. We store the classified value back into a label and store the result as:

   - `Irrigation_potential` ∈ {Low, Medium, High, Very High}


### 4. Combined_PUE_potential 

**Inputs:**

- `Cooling_potential` 
- `Irrigation_potential`

**Rationale:**

A combined productive-use potential indicator is calculated from the cooling and irrigation potential classifications. Both indicators use the following ordinal classification:

| Classification | Numerical score |
|---|---:|
| Low | 1 |
| Medium | 2 |
| High | 3 |
| Very High | 4 |

The cooling and irrigation classifications are first converted to their corresponding numerical scores. The combined potential score is then calculated as the arithmetic mean of the two indicators. For example:

| Cooling potential | Irrigation potential | Combined score |
|---|---|---:|
| Low | Medium | 1.5 |
| Medium | High | 2.5 |
| High | Very High | 3.5 |
| Very High | Very High | 4.0 |

#### Treatment of missing values

Missing values are ignored when calculating the average:

- When both cooling and irrigation values are available, the combined score is their average.
- When only one indicator is available, the combined score is equal to the available indicator's score.
- When both indicators are missing, the combined score remains missing.
- Empty strings and strings containing only spaces are treated as missing values.

The final column is stored as a continuous numerical score between 1 and 4

   - `Combined_PUE_potential` ∈ {1, 1.5, 2, 2.5, 3, 3.5, 4}

---

### Why this approach is useful

- **Evidence-based**: thresholds are derived from the actual data distribution (quantiles), not arbitrary cut-offs.
- **Interpretable**: labels (Low/Medium/High/Very High) are intuitive for planners and developers.
- **Sector-neutral**: cooling potential respects the strongest of all cooling demand channels; irrigation potential respects the strong water needs
- **Decision-ready**: `Combined_PUE_potential` provides a single, easy-to-use indicator to filter and prioritise settlements related to their PUE potential

### Setting column names, class definitions and functions needed

```python
## Columns for accessibility potential
access_cols = [
    "Fresh Markets Accessibility",
    "National Market Accessibility",
    "Export Market Accessibility",
]

cool_cols = [
    "Ag Cooling Demand ALL Markets",
    "Fish Cooling Demand ALL Markets",
]

irr_cols = [
    "Irrigation: months with (moderate + critical) need",
    "Irrigation: months with (critical) need",
]

# Evidence-based class labels (quartiles)
class_labels = ["Low", "Medium", "High", "Very High"]
class_to_int = {lab: i for i, lab in enumerate(class_labels)}  # Low=0 ... Very High=3
int_to_class = {i: lab for lab, i in class_to_int.items()}
```

```python
def classify_quantiles(series, labels=class_labels):
    """
    Classify a numeric pandas Series into quantile-based classes.
    By default, splits into 4 quantiles with labels:
    Low, Medium, High, Very High.
    """
    # Work only on non-null values
    non_null = series.dropna()

    if non_null.nunique() == 0:
        # Everything is the same or all NaN; return all NaN classes
        return pd.Series(index=series.index, dtype="object")

    # Use qcut to split into quantiles.
    # duplicates='drop' avoids errors when there are too few unique values.
    cats = pd.qcut(
        non_null,
        q=len(labels),
        labels=labels,
        duplicates="drop"
    )

    # Reinsert into full index, keeping NaNs where original data was NaN
    out = pd.Series(index=series.index, dtype="object")
    out.loc[non_null.index] = cats.astype(str)

    return out
```

```python
def class_to_score(label):
    """Convert class label to ordinal score; NaNs stay NaN."""
    if pd.isna(label):
        return None
    return class_to_int[label]
```

#### 1. Calculating Accessibility_potential

```python
# Create class columns for each cooling index
access_class_cols = []
for col in access_cols:
    class_col = f"{col} - class"
    settles_gdf_analyzed[class_col] = classify_quantiles(settles_gdf_analyzed[col])
    access_class_cols.append(class_col)
```

#### 2. Calculating Cooling_potential

```python
# Safety check: ensure cooling columns exist
missing_cool = [c for c in cool_cols if c not in settles_gdf_analyzed.columns]
if missing_cool:
    raise ValueError(f"Missing cooling columns in dataframe: {missing_cool}")

# Create class columns for each cooling index
cool_class_cols = []
for col in cool_cols:
    class_col = f"{col} - class"
    settles_gdf_analyzed[class_col] = classify_quantiles(settles_gdf_analyzed[col])
    cool_class_cols.append(class_col)

# Convert class labels to ordinal integers and pick highest per row
cool_class_int = settles_gdf_analyzed[cool_class_cols].apply(
    lambda row: max(
        class_to_int[val] for val in row.dropna()
    ) if row.notna().any() else None,
    axis=1
)

# Map back to labels
settles_gdf_analyzed["Cooling_potential"] = cool_class_int.map(int_to_class)
```

#### 3. Calculating Irrigation_potential

```python
# Safety check: ensure cooling columns exist
missing_irr = [c for c in irr_cols if c not in settles_gdf_analyzed.columns]
if missing_cool:
    raise ValueError(f"Missing cooling columns in dataframe: {missing_irr}")

# Create class columns for each cooling index
irr_class_cols = []
for col in irr_cols:
    class_col = f"{col} - class"
    settles_gdf_analyzed[class_col] = classify_quantiles(settles_gdf_analyzed[col])
    irr_class_cols.append(class_col)

# Assign labels to new column
settles_gdf_analyzed["Irrigation_potential"] = settles_gdf_analyzed["Irrigation: months with (critical) need - class"]
```

#### 3. Calculating Combined_PUE_potential

```python
potential_cols = ["Cooling_potential", "Irrigation_potential"]

class_to_int = {
    "Low": 1,
    "Medium": 2,
    "High": 3,
    "Very High": 4
}

int_to_class = {value: key for key, value in class_to_int.items()}

# Change to True only when rounding and categorical output are desired
apply_rounding = False

# Clean and convert the original classifications to numbers
potential_scores = settles_gdf_analyzed[potential_cols].apply(
    lambda col: (
        col.astype("string")
        .str.strip()
        .replace("", pd.NA)
        .map(class_to_int)
    )
)

# Calculate the average, ignoring a null when the other value exists
combined_score = potential_scores.mean(axis=1, skipna=True)

if apply_rounding:
    # Round halves upward: 1.5 → 2, 2.5 → 3, 3.5 → 4
    rounded_score = np.floor(combined_score + 0.5)

    settles_gdf_analyzed["Combined_PUE_potential"] = (
        rounded_score.map(int_to_class).astype("string")
    )
else:
    # Keep the original average as a numerical score
    settles_gdf_analyzed["Combined_PUE_potential"] = (
        combined_score.astype("Float64")
    )
```

```python
ag_class_col   = 'Ag Cooling Demand ALL Markets - class'
fish_class_col = 'Fish Cooling Demand ALL Markets - class'
irri_class_col = 'Irrigation_potential'

# --- Scenario 1: Post-Harvest Cooling (Ag) ---
postharvest_df = settles_gdf_analyzed.copy()
postharvest_df['PUE_potential'] = postharvest_df[ag_class_col]
#postharvest_df = postharvest_df.dropna(subset=['PUE_potential'])

# --- Scenario 2: Fishing Cooling ---
fishing_df = settles_gdf_analyzed.copy()
fishing_df['PUE_potential'] = fishing_df[fish_class_col]
#fishing_df = fishing_df.dropna(subset=['PUE_potential'])

# --- Scenario 3: Irrigation ---
irrigation_df = settles_gdf_analyzed.copy()
irrigation_df['PUE_potential'] = irrigation_df[irri_class_col]
#irrigation_df = irrigation_df.dropna(subset=['PUE_potential'])

# --- UTF-8 compatibility and null filling ---
def prepare_for_export(df):
    for col in df.columns:
        if col == 'geometry':
            continue
        if df[col].dtype == 'object':
            df[col] = df[col].fillna('').apply(
                lambda x: x.encode('utf-8', errors='replace').decode('utf-8')
                if isinstance(x, str) else x
            )
        else:
            df[col] = df[col].fillna(0)
    return df

postharvest_df = prepare_for_export(postharvest_df)
fishing_df     = prepare_for_export(fishing_df)
irrigation_df     = prepare_for_export(irrigation_df)
```
